## Exercise One — Order Processing as a Function

Encapsulate the discount tiers **and** the math inside one function that **returns** `(subtotal, total_with_tax, final_total)`. The caller handles input and formatting.

**Assumptions (carried from earlier in Week 1):** tax rate `0.075`; discount applied to the tax-inclusive total; tiers 0% / 10% (50–99) / 25% (100+).

```python
TAX_RATE = 0.075


def calculate_order(quantity, unit_cost):
    """Return (subtotal, total_with_tax, final_total) for an order.

    The discount tier is derived from quantity and applied after tax:
        quantity >= 100 -> 25%,  quantity >= 50 -> 10%,  else 0%.
    """
    if quantity >= 100:
        discount_pct = 25
    elif quantity >= 50:
        discount_pct = 10
    else:
        discount_pct = 0

    subtotal = quantity * unit_cost
    total_with_tax = subtotal * (1 + TAX_RATE)
    final_total = total_with_tax * (1 - discount_pct / 100)
    return subtotal, total_with_tax, final_total


def main():
    part_number = input('Part number: ')
    quantity = int(input('Quantity: '))
    unit_cost = float(input('Unit cost: '))

    subtotal, total_with_tax, final_total = calculate_order(quantity, unit_cost)

    print('--- Order Detail ---')
    print(f'Part number:        {part_number}')
    print(f'Quantity:           {quantity}')
    print(f'Subtotal:           ${subtotal:,.2f}')
    print(f'Total (with tax):   ${total_with_tax:,.2f}')
    print(f'Final total:        ${final_total:,.2f}')


# Uncomment to run interactively:
main()
```

**Sample run (no typing needed)** — calling the function directly and unpacking its tuple:

```python
for qty in (10, 60, 120):
    subtotal, total_with_tax, final_total = calculate_order(qty, 1.99)
    print(f'qty={qty:>3}  subtotal=${subtotal:>7,.2f}  with_tax=${total_with_tax:>7,.2f}  final=${final_total:>7,.2f}')
```

> The function **returns** values rather than printing them — so the same calculation can be reused, tested, or fed into other code. Watch for solutions that `print` inside the function and return nothing; that works on screen but defeats the exercise.

## Exercise Two — `circleinfo`

Return area (`Pi * r * r`) and circumference (`2 * Pi * r`) for a given radius.

> **Note the hint correction:** use `math.pi` — a **constant**, no parentheses. `math.pi()` raises `TypeError: 'float' object is not callable`.

```python
import math


def circleinfo(radius):
    """Return (area, circumference) for a circle of the given radius."""
    area = math.pi * radius * radius
    circumference = 2 * math.pi * radius
    return area, circumference


def main():
    radius = float(input('Radius: '))
    area, circumference = circleinfo(radius)
    print(f'Area:          {area:.4f}')
    print(f'Circumference: {circumference:.4f}')


# Uncomment to run interactively:
main()
```

**Sample run (no typing needed):**

```python
for r in (1, 5, 10):
    area, circumference = circleinfo(r)
    print(f'r={r:>2}  area={area:>8.4f}  circumference={circumference:>8.4f}')
```

> **Watch for:** `math.pi()` (the literal hint bug — raises `TypeError`), and computing the value once but only returning one of the two results. `area = math.pi * radius ** 2` is an equally good way to write the area.
