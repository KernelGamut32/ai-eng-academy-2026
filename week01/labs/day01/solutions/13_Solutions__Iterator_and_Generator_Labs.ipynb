# Solution Key — Iterables, Iterators, and Generators
## Iterator Lab & Generator Lab

The second lab is the first one rewritten as a generator — same behavior, far less code. The sample runs seed `random` so the output is reproducible here; **the order will differ every run without a seed**, which is the whole point.

---
## Lab 1 — `MyRandomIterator`

Take an iterable; each `next()` returns a *random* element not yet returned. When all elements have been handed out, raise `StopIteration`.

```python
import random


class MyRandomIterator:
    """Yield the elements of an iterable in random order, once each."""
    def __init__(self, iterable):
        self._items = list(iterable)   # copy so we can consume it destructively

    def __iter__(self):
        return self                    # the object is its own iterator

    def __next__(self):
        if not self._items:            # nothing left -> we're done
            raise StopIteration
        i = random.randrange(len(self._items))
        return self._items.pop(i)      # remove & return a random element


random.seed(7)
print('as a list :', list(MyRandomIterator([1, 2, 3, 4, 5])))

random.seed(7)
print('in a loop :', end=' ')
for value in MyRandomIterator(['a', 'b', 'c', 'd']):
    print(value, end=' ')
print()
```

> **Note:** the two requirements are (1) `__next__` returns elements *without repeats* and (2) it `raise`s `StopIteration` when exhausted. `pop(random index)` handles both — the list shrinks each call, and empties exactly when every element has been returned. Watch for solutions that pick a random element but don't remove it (infinite / repeating) or that return `None` instead of raising at the end (a `for` loop would never stop). `__iter__` returning `self` is what lets it work directly in a `for` loop.

## Lab 2 — The Same Thing as a Generator

A generator function makes this dramatically shorter: no class, no `__iter__`/`__next__`, and no explicit `raise` — **falling off the end of the function raises `StopIteration` for you**.

```python
import random


def random_generator(iterable):
    """Yield the elements of an iterable in random order, once each."""
    items = list(iterable)
    while items:                       # loop ends when the list is empty...
        i = random.randrange(len(items))
        yield items.pop(i)
    # ...and falling off the end here ends the generator (StopIteration)


random.seed(7)
print('as a list :', list(random_generator([1, 2, 3, 4, 5])))

random.seed(7)
print('in a loop :', end=' ')
for value in random_generator(['a', 'b', 'c', 'd']):
    print(value, end=' ')
print()
```

> **The takeaway to highlight:** the generator is ~6 lines versus a ~12-line class and produces identical behavior — same seeded output as Lab 1. The `while items:` condition replaces the explicit `if not self._items: raise StopIteration`; the generator protocol handles termination automatically. This is exactly the "elegant, succinct syntax" the notebook promised generators would give over hand-written iterator classes.
