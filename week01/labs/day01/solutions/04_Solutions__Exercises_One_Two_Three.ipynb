## Exercise One — Square Root via Newton's Method

Start the guess at `1.0`, apply `guess = guess - (guess**2 - x) / (2 * guess)` ten times, printing each guess. The EXTRA wraps input in a validation loop that re-prompts until the value is positive.

```python
def read_positive_float(prompt):
    """Re-prompt until the user enters a number > 0 (the EXTRA)."""
    while True:
        value = float(input(prompt))
        if value > 0:
            return value
        print('Please enter a positive number.')


def newton_sqrt(x, iterations=10):
    """Approximate sqrt(x) with Newton's method, printing each guess."""
    guess = 1.0
    for i in range(1, iterations + 1):
        guess = guess - (guess ** 2 - x) / (2 * guess)
        print(f'Iteration {i:>2}: {guess}')
    return guess


def main():
    x = read_positive_float('Enter a positive number: ')
    result = newton_sqrt(x)
    print(f'Approximate square root: {result}')


# Uncomment to run interactively:
main()
```

**Sample run (no typing needed)** — `x = 25`, converges to 5.0:

```python
newton_sqrt(25)
```

> **The point:** the guess homes in fast (essentially exact by ~iteration 7 for 25). Watch forgetting to update `guess` each pass or using the *previous* guess on both sides of the formula incorrectly. Dividing by `2 * guess` is why the starting guess must be non-zero — the `1.0` start sidesteps a divide-by-zero.

## Exercise Two — Factorial

Compute `n!` with a loop. The EXTRA re-prompts until the input is a non-negative integer. Recall `0! == 1`.

```python
def read_non_negative_int(prompt):
    """Re-prompt until the user enters an integer >= 0 (the EXTRA)."""
    while True:
        value = int(input(prompt))
        if value >= 0:
            return value
        print('Please enter a non-negative integer.')


def factorial(n):
    """Return n! using a loop. factorial(0) == 1 (empty product)."""
    result = 1
    for i in range(2, n + 1):   # range is empty for n in (0, 1) -> result stays 1
        result *= i
    return result


def main():
    n = read_non_negative_int('Enter a non-negative integer: ')
    print(f'{n}! = {factorial(n)}')


# Uncomment to run interactively:
main()
```

**Sample run (no typing needed):**

```python
for n in (0, 1, 5, 10):
    print(f'{n}! = {factorial(n)}')
```

> **The point:** the `0!` and `1!` edge cases. Starting `result = 1` and ranging from 2 handles both for free. Watch for solutions that start the loop at 1 (harmless but redundant) or at 0 (multiplies everything by zero — a classic bug).

## Exercise Three — Collatz Conjecture

Accept an integer `>= 1` (return `False` if `< 1`). Even → halve; odd → `3n + 1`. Stop at 1. Print each step and the total count, then return `True`.

```python
def collatz(n):
    """Run the Collatz sequence from n. Print each value and the step count.

    Returns False if n < 1, otherwise True once the sequence reaches 1.
    """
    if n < 1:
        return False

    steps = 0
    print(n)
    while n != 1:
        if n % 2 == 0:
            n = n // 2          # integer division keeps n an int
        else:
            n = 3 * n + 1
        print(n)
        steps += 1

    print(f'Reached 1 in {steps} steps')
    return True


# Uncomment to run interactively:
n = int(input('Enter an integer >= 1: '))
collatz(n)
```

**Sample runs (no typing needed):**

```python
print('--- collatz(6) ---')
print('returned:', collatz(6))
print()
print('--- collatz(7) ---')
print('returned:', collatz(7))
print()
print('--- collatz(0) ---')
print('returned:', collatz(0))   # < 1, returns False with no sequence
```

> **Watch for:** using `/` instead of `//` for the even case — `/` yields a float, so you'd print `10.0, 5.0, ...` and the `n != 1` test compares against `1.0`. It still terminates, but integer division is the right tool. Also confirm the early `return False` for `n < 1` happens **before** any printing. `collatz(1)` should print just `1` and report 0 steps.
