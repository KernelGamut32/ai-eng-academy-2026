## Exercise One — `minmaxavg`

Return min, max, and average of a list of floats. Built-ins `min`, `max`, and `sum` (all shown in the lesson) make this a one-liner; a manual loop is equally acceptable.

```python
def minmaxavg(values):
    """Return (minimum, maximum, average) of a non-empty list of numbers."""
    return min(values), max(values), sum(values) / len(values)


def main():
    count = int(input('How many values? '))
    values = []
    for i in range(count):
        values.append(float(input(f'Value {i + 1}: ')))

    low, high, avg = minmaxavg(values)
    print(f'Min: {low:.2f}   Max: {high:.2f}   Avg: {avg:.2f}')


# Uncomment to run interactively:
main()
```

**Sample run (no typing needed):**

```python
low, high, avg = minmaxavg([3.5, 1.2, 9.8, 4.4])
print(f'Min: {low:.2f}   Max: {high:.2f}   Avg: {avg:.2f}')
```

> **Watch for:** dividing by `len` *before* summing, or building the list with strings instead of `float(...)` (then `min`/`max` compare lexicographically and `sum` fails). Empty list is the one crash case — `min([])` raises `ValueError`; leaving out of scope for this lab.

## Exercise Two — `fibonacci`

Return a list of the first `count` Fibonacci numbers, starting `1, 1, 2, 3, 5, ...`

```python
def fibonacci(count):
    """Return a list of the first `count` Fibonacci numbers (1, 1, 2, 3, 5, ...)."""
    fibs = [1, 1]
    while len(fibs) < count:
        fibs.append(fibs[-1] + fibs[-2])
    return fibs[:count]   # trim in case count < 2


def main():
    count = int(input('How many Fibonacci numbers? '))
    print(fibonacci(count))


# Uncomment to run interactively:
main()
```

**Sample run (no typing needed):**

```python
print(fibonacci(10))
```

> **Watch for:** the seed. Starting `[1, 1]` matches the exercise's sequence; a `[0, 1]` seed produces `0, 1, 1, 2, ...` which is a different (also valid) convention — flag the mismatch if their output starts with 0. The `[:count]` trim keeps small requests from over-returning.

## Exercise Three — Roman → Arabic

The **subtractive** algorithm handles both the basic case and the EXTRA in one pass: walk left to right; if a symbol's value is **less than the next** symbol's, subtract it, otherwise add it.

Per the exercise, `MCLX -> 1160` and `MCMLCIX -> 1959`. (`MCMLCIX` isn't a canonical numeral, but the left-to-right subtractive rule yields 1959 exactly as the prompt specifies.)

```python
ROMAN_VALUES = {'M': 1000, 'D': 500, 'C': 100, 'L': 50, 'X': 10, 'V': 5, 'I': 1}


def roman_to_arabic(roman):
    """Convert an (upper-case) Roman numeral string to its integer value."""
    total = 0
    for i, symbol in enumerate(roman):
        value = ROMAN_VALUES[symbol]
        # If a smaller value sits before a larger one, it is subtracted.
        if i + 1 < len(roman) and ROMAN_VALUES[roman[i + 1]] > value:
            total -= value
        else:
            total += value
    return total


def main():
    roman = input('Enter a Roman numeral: ').upper()
    print(f'{roman} = {roman_to_arabic(roman)}')


# Uncomment to run interactively:
main()
```

**Sample run (no typing needed):**

```python
for r in ('MCLX', 'MCMLCIX', 'MCMLIX', 'MMXXIV'):
    print(f'{r:>10} -> {roman_to_arabic(r)}')
```

## Exercise Four — Dictionary-Dispatched Calculator

Map operator strings to functions in a dict, parse `"2 + 4"`, look up the function, call it. Per the spec each function returns an **integer**, so division uses `//` (integer division).

```python
def addition(a, b):
    return a + b

def subtraction(a, b):
    return a - b

def multiplication(a, b):
    return a * b

def division(a, b):
    return a // b        # integer result per the spec; use / for true division


# The dict maps a symbol -> the function object itself (functions are objects).
OPERATIONS = {'+': addition, '-': subtraction, '*': multiplication, '/': division}


def calculate(line):
    """Parse a "<int> <op> <int>" line and return the integer result."""
    left, operator, right = line.split()
    func = OPERATIONS[operator]
    return func(int(left), int(right))


def main():
    line = input('Enter a calculation (e.g. 2 + 4): ')
    print(f'{line} = {calculate(line)}')


# Uncomment to run interactively:
main()
```

**Sample run (no typing needed):**

```python
for line in ('2 + 4', '10 - 3', '6 * 7', '20 / 6'):
    print(f'{line:>8} = {calculate(line)}')
```

> **Key point:** the dict stores **function objects**, not strings or results — `OPERATIONS['+']` *is* `addition`, and `OPERATIONS[op](a, b)` calls it. This is the same dispatch-table pattern we might see later for routing requests or model calls. 
>
> **Watch for:** `line.split()` assumes single spaces around the operator (`"2+4"` would fail) — fine for this lab. `20 / 6 -> 3` because `//` truncates; if you expect `3.33`, that's the `//` vs `/` distinction. Division by zero raises `ZeroDivisionError`; guarding it is a reasonable stretch but not required.
