## Exercise One — User Profile

Prompt for first name, last name, age, years of experience, and job title, then print the profile in an organized format.

```python
def show_profile(first, last, age, years_experience, title):
    """Print a user profile in a clean, aligned format."""
    print('--- User Profile ---')
    print(f'Name:        {first} {last}')
    print(f'Age:         {age}')
    print(f'Title:       {title}')
    print(f'Experience:  {years_experience} year(s) in current role')


def main():
    first = input('First name: ')
    last = input('Last name: ')
    age = int(input('Age: '))
    years_experience = int(input('Years of experience in current role: '))
    title = input('Job title: ')
    show_profile(first, last, age, years_experience, title)


# Uncomment to run interactively:
main()
```

**Sample run (no typing needed):**

```python
show_profile('Dana', 'Rivera', 34, 8, 'Senior Software Engineer')
```

## Exercise Two — Order Processing

Prompt for part number, quantity, unit cost, and discount, then calculate and print subtotal, total including tax, and final total after discount.

**Assumptions (stated, since the prompt is open):**
* **Tax rate** is a fixed `0.075` (7.5%), matching the demo earlier in the lesson.
* **Discount** is entered as a whole-number **percent** (e.g., `10` = 10% off).
* **Order of operations:** subtotal → add tax → then apply discount to the tax-inclusive total. Applying the discount *before* tax is equally defensible; just be consistent and say which you chose.

```python
TAX_RATE = 0.075  # 7.5% — single source of truth, easy to change


def process_order(part_number, quantity, unit_cost, discount_pct):
    """Compute and print an order summary.

    discount_pct is a whole-number percent (10 means 10%).
    Tax is applied first, then the discount is applied to the taxed total.
    """
    subtotal = quantity * unit_cost
    total_with_tax = subtotal * (1 + TAX_RATE)
    final_total = total_with_tax * (1 - discount_pct / 100)

    print('--- Order Detail ---')
    print(f'Part number:        {part_number}')
    print(f'Quantity:           {quantity}')
    print(f'Unit cost:          ${unit_cost:,.2f}')
    print(f'Subtotal:           ${subtotal:,.2f}')
    print(f'Total (with tax):   ${total_with_tax:,.2f}')
    print(f'Discount:           {discount_pct:g}%')
    print(f'Final total:        ${final_total:,.2f}')


def main():
    part_number = input('Part number: ')
    quantity = int(input('Quantity: '))
    unit_cost = float(input('Unit cost: '))
    discount_pct = float(input('Discount (%): '))
    process_order(part_number, quantity, unit_cost, discount_pct)


# Uncomment to run interactively:
main()
```

**Sample run (no typing needed):**

```python
process_order('HW-4471', 12, 1.99, 10)
```
