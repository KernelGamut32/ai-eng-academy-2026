## Exercise One — Word Frequency Count

Count occurrences of each word using a **dictionary** keyed by word. Case-fold so `The` and `the` are one word; the EXTRA strips punctuation so `night,` and `night` match.

```python
import re


def count_words(filename):
    """Return {word: count} for a file, case-folded and punctuation-stripped."""
    counts = {}
    with open(filename) as f:
        for line in f:
            for token in line.split():
                word = token.lower()
                word = re.sub(r'[^a-z0-9]', '', word)   # EXTRA: drop punctuation
                if word:                                # skip tokens that were all punctuation
                    counts[word] = counts.get(word, 0) + 1
    return counts


def main(filename='../supporting-materials/poem.txt', top=10):
    counts = count_words(filename)
    # Sort by count (desc), then alphabetically for stable, readable output.
    ranked = sorted(counts.items(), key=lambda kv: (-kv[1], kv[0]))
    print(f'{len(counts)} distinct words. Top {top}:')
    for word, n in ranked[:top]:
        print(f'{word:>12} : {n}')


main()
```

> **The point:** `counts.get(word, 0) + 1` is the idiomatic dict-as-counter pattern from the datatypes lesson. `collections.Counter` does the same in one line and is worth mentioning as the production tool, but the exercise specifically asks for a dictionary. 
>
> **Watch for:** counting before lower-casing (then `The` and `the` split), and the apostrophe trade-off — stripping every non-alphanumeric turns `don't` into `dont`. Fine for this lab but take note. To run on a different text file, just call `main('text-file.txt')`.

## Exercise Two — Pluralization

Apply the four rules in order. The character checks are done on a lower-cased copy so the rules work regardless of input case; the suffix is appended to the original word.

```python
VOWELS = 'aeiou'
# Consonants after which a final 'h' is NOT pluralized with 'es' (moth -> moths).
H_NO_ES = 'dgkprt'


def pluralize(word):
    """Return the plural of a single word using the four rules."""
    w = word.lower()

    if w.endswith(('s', 'x', 'z')):              # ax -> axes, loss -> losses
        return word + 'es'

    if w.endswith('h'):                          # match -> matches, moth -> moths
        if len(w) >= 2 and w[-2] not in VOWELS and w[-2] not in H_NO_ES:
            return word + 'es'
        return word + 's'

    if w.endswith('y'):                          # baby -> babies, boy -> boys
        if len(w) >= 2 and w[-2] not in VOWELS:
            return word[:-1] + 'ies'
        return word + 's'

    return word + 's'                            # default: cat -> cats


def main():
    word = input('Enter a word to pluralize: ')
    print(f'{word} => {pluralize(word)}')


# Uncomment to run interactively:
main()
```

**Sample run (no typing needed)** — covers every rule branch:

```python
checks = ['ax', 'loss', 'buzz', 'moth', 'match', 'dish', 'graph',
          'baby', 'boy', 'key', 'box', 'cat']
for word in checks:
    print(f'{word:>7} => {pluralize(word)}')
```

> **The rule that trips people up** is the `h` case: it adds `es` only when the letter *before* the `h` is neither a vowel nor one of `d g k p r t`. So `match -> matches` (c before h) but `moth -> moths` (t before h) and `graph -> graphs` (p before h). Check those three explicitly. English has plenty of real exceptions (`mouse`, `child`, `goose`) — out of scope; these four rules are the assignment.
