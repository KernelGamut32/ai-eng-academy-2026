## Lab: Class Methods

Track instances with two **class variables** and expose them via **class methods**:
* `allnames()` → the names of objects that **currently exist** (maintained with `__del__`)
* `count()` → the number of objects **ever created** (only ever increments)

```python
class Critter:
    _existing = []   # class var: names of instances alive right now
    _ever = 0        # class var: how many have ever been created

    def __init__(self, name):
        self.name = name
        Critter._existing.append(name)
        Critter._ever += 1

    def __del__(self):
        # Remove this instance's name when it is garbage-collected.
        if self.name in Critter._existing:
            Critter._existing.remove(self.name)

    @classmethod
    def allnames(cls):
        return list(cls._existing)   # return a copy, not the live list

    @classmethod
    def count(cls):
        return cls._ever
```

**Sample run:**

```python
a = Critter('Ann')
b = Critter('Bob')
c = Critter('Cy')
print('after creating 3 ->', Critter.allnames(), '| ever:', Critter.count())

del b   # Bob is gone
print('after deleting Bob ->', Critter.allnames(), '| ever:', Critter.count())
```

> **Notes & gotchas:**
> * `count()` must use a separate counter that **never decrements** — a common bug is returning `len(_existing)`, which would drop when objects are deleted (that's `allnames`, not `count`).
> * The class vars are referenced as `Critter._existing` / `Critter._ever` (or `cls.` in the classmethods), not `self._existing = ...`, which would shadow them per-instance.
> * `__del__` fires on garbage collection. With CPython's reference counting, an explicit `del` that drops the last reference runs it immediately — but in a notebook, watch out: Jupyter keeps references in `Out[...]`, `_`, `__`, etc., so an object echoed as a cell's result may linger past a `del`. Mention this; it's the same subtlety the `Example`/`__how_many` cells illustrate.
> * Duplicate names: removing by value (`_existing.remove(self.name)`) deletes the first match, which is fine unless two instances share a name. Tracking the objects themselves (or ids) is a reasonable, more robust variation.
