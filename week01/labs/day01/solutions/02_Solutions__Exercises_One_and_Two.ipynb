## Exercise One — Order Processing with a Tiered Discount

Build on the order program, but **derive** the discount from quantity instead of prompting:
* `quantity >= 100` → 25%
* `quantity >= 50` (and `< 100`) → 10%
* otherwise → 0%

**Assumptions (carried over from the Week 1 lesson):** tax rate is `0.075`; the discount is applied to the **tax-inclusive** total. Any consistent, stated choice is acceptable.

```python
TAX_RATE = 0.075


def discount_for(quantity):
    """Return the discount percent (whole number) for a given quantity.

    Checking the highest threshold first keeps the logic clean: once we know
    quantity >= 100 is false, 'quantity >= 50' already implies quantity < 100.
    """
    if quantity >= 100:
        return 25
    elif quantity >= 50:
        return 10
    else:
        return 0


def process_order(part_number, quantity, unit_cost):
    """Compute and print an order summary with an auto-derived discount."""
    discount_pct = discount_for(quantity)
    subtotal = quantity * unit_cost
    total_with_tax = subtotal * (1 + TAX_RATE)
    final_total = total_with_tax * (1 - discount_pct / 100)

    print('--- Order Detail ---')
    print(f'Part number:        {part_number}')
    print(f'Quantity:           {quantity}')
    print(f'Unit cost:          ${unit_cost:,.2f}')
    print(f'Subtotal:           ${subtotal:,.2f}')
    print(f'Total (with tax):   ${total_with_tax:,.2f}')
    print(f'Discount:           {discount_pct:g}%')
    print(f'Final total:        ${final_total:,.2f}')


def main():
    part_number = input('Part number: ')
    quantity = int(input('Quantity: '))
    unit_cost = float(input('Unit cost: '))
    process_order(part_number, quantity, unit_cost)


# Uncomment to run interactively:
main()
```

**Sample runs (no typing needed)** — one per discount tier:

```python
process_order('HW-4471', 10, 1.99)    # below 50  -> 0%
print()
process_order('HW-4471', 60, 1.99)    # 50-99     -> 10%
print()
process_order('HW-4471', 120, 1.99)   # 100+      -> 25%
```

> **Watch for:** the classic off-by-one on the boundary — `>= 50` vs `> 50`, and `>= 100` vs `> 100`. Quantities of exactly 50 and exactly 100 are the cases to check. Also order the branches so the ranges can't overlap (highest threshold first).

## Exercise Two — FizzBuzz

Read an integer and print `fizz` (÷3), `buzz` (÷5), `fizzbuzz` (÷ both), else the number itself.

```python
def fizzbuzz(n):
    """Return the FizzBuzz string for a single integer."""
    if n % 3 == 0 and n % 5 == 0:   # divisible by BOTH -> check first
        return 'fizzbuzz'
    elif n % 3 == 0:
        return 'fizz'
    elif n % 5 == 0:
        return 'buzz'
    else:
        return str(n)


def main():
    n = int(input('Enter an integer: '))
    print(fizzbuzz(n))


# Uncomment to run interactively:
main()
```

**Sample run (no typing needed):**

```python
for n in [9, 10, 30, 7]:
    print(f'{n:>3} -> {fizzbuzz(n)}')
```

> **The key gotcha:** the *both* case must be tested **before** the individual 3 and 5 checks. Writing `if n % 3 == 0` first will never reach `fizzbuzz` and will print `fizz` for 15. Equivalent clean alternatives: test `n % 15 == 0` first, or build the word by concatenation (`'fizz' * (n % 3 == 0) + 'buzz' * (n % 5 == 0)` and fall back to the number when empty).
