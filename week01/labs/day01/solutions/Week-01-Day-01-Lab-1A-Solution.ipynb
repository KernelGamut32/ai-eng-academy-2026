## Step 01 — Define a function

A small **typed** text-cleaning function over an inline list — a preview of the data-transform work that fills every AI pipeline. The list comprehension does the strip / lower-case / drop-empty in one readable pass.

```python
def clean_tokens(tokens: list[str]) -> list[str]:
    """Normalize a list of raw token strings.

    Strip surrounding whitespace, lower-case each token, and drop any
    token that is empty after stripping.

    Args:
        tokens: Raw input strings.

    Returns:
        Cleaned, non-empty tokens.
    """
    return [t.strip().lower() for t in tokens if t.strip()]
```

```python
# Call the function on a small inline list and look at the result.
raw = ["  The ", "CAT", "  ", "sat", "ON", ""]
clean_tokens(raw)
```

**Expected output:** `['the', 'cat', 'sat', 'on']`

Note the two whitespace-only / empty entries (`"  "` and `""`) are dropped by the `if t.strip()` filter — a good place to ask learners *why* filtering on the stripped value (not the raw value) matters.

## Step 02 — Time two approaches with `%timeit`

`%timeit` is a **line magic**: it benchmarks a single expression and auto-tunes the number of loops and repeats for a stable estimate. To time the hand-written loop with a *line* magic we wrap it in `loop_sum` and time the call.

```python
def loop_sum(n: int) -> int:
    """Sum 0..n-1 with an explicit Python loop."""
    total = 0
    for i in range(n):
        total += i
    return total
```

```python
%timeit loop_sum(1_000_000)
```

```python
%timeit sum(range(1_000_000))
```

**What to expect:** the built-in `sum()` runs clearly faster than `loop_sum` (commonly ~2–10×, machine- and build-dependent). If the two are suspiciously close, check that nothing else is hammering the CPU.

## Step 03 — Explain it (Markdown)

> ✍️ **Why the built-in is faster:** `sum()` runs its accumulation loop in optimized C inside the interpreter, so it avoids the per-iteration Python bytecode overhead — interpreter dispatch and a fresh integer object on every pass — that the hand-written `for` loop pays.

*Teaching note:* this is the same intuition behind preferring vectorized NumPy/pandas operations over Python loops — which is exactly where Day 2 goes.

## 🚀 Stretch — `%%timeit` cell magic

`%%timeit` times the **entire cell**, so it benchmarks the multi-line loop directly — no helper function needed. It must be the very first line of the cell.

```python
%%timeit
total = 0
for i in range(1_000_000):
    total += i
```

> ✍️ **Line vs. cell magic:** `%timeit` is a **line magic** (single `%`) — it benchmarks one expression on the rest of that line. `%%timeit` is a **cell magic** (double `%`) — it must be the first line of the cell and benchmarks the whole cell body, which is the tool for multi-line blocks. Both auto-tune loops and repeats for a stable estimate. Reach for the line magic for a quick one-liner; reach for the cell magic to time a multi-step block.

*Sanity check:* the `%%timeit` inline-loop number should land close to the `%timeit loop_sum(1_000_000)` number from Step 02 — they're doing the same work, just measured two different ways.

## 👩‍🏫 Notes — likely pitfalls

Common things to watch for (good discussion openers):

- **Timing a loop with a line magic.** You can't `%timeit` a multi-line loop directly — users either wrap it in a function (as we did) or discover they need `%%timeit`. Either realization is the point.
- **`%%timeit` not on the first line.** If any code (even a comment on line 1 before the magic) precedes `%%timeit`, it errors. Quick fix, good muscle memory.
- **Comprehension for side effects.** Watch for `[print(t) for t in tokens]` — a comprehension built to throw away its result is a loop in disguise. Comprehensions build collections; use a plain loop for side effects.
- **Filtering on the raw value.** `if t` instead of `if t.strip()` keeps whitespace-only strings. Note that the `"  "` entry survives.
- **Restart-and-Run-All skipped.** Focue on reproducible — the hidden-state trap. Make the restart non-negotiable.
