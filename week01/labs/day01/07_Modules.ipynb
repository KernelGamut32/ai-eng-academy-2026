```python
import mymodule
mymodule.dummy() # must preface identifiers with module name
```

> **Note:** importing a module **runs its top-level code once** — that's why the line above prints `__name__ = mymodule`. On import, `__name__` is the module's name (`mymodule`), so the `if __name__ == '__main__'` block is skipped (no `success`). Contrast with running it directly (cells above), where `__name__` is `__main__`. Re-running the import won't print again; Python caches modules in `sys.modules`.
