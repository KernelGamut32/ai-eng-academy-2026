## Step 02 — Time two approaches with `%timeit`

`%timeit` is a **line magic**: it times a single expression and automatically runs it many times for a statistically stable estimate.

`loop_sum` is provided for you below. Your task is to **time both approaches** and compare:
- the hand-written loop: `loop_sum(1_000_000)`
- Python's built-in: `sum(range(1_000_000))`

```python
def loop_sum(n: int) -> int:
    """Sum 0..n-1 with an explicit Python loop."""
    total = 0
    for i in range(n):
        total += i
    return total
```
